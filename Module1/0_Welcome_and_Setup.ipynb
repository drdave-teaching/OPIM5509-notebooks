## 2. Where did it go?!

You clicked `File > Save a copy in Drive`, the tab reloaded, the title now says **Copy of ...** - and the file is nowhere you were looking.

It went here:

```
My Drive / Colab Notebooks / Copy of 0_Welcome_and_Setup.ipynb
```

**Always that folder.** Colab creates `Colab Notebooks` at the top level of your Drive the first time you save, and from then on it drops every copy there without asking. It does not matter that you opened this notebook from GitHub - the copy does not land anywhere near GitHub, and it does not land in whatever folder you happened to be browsing.

Two shortcuts worth knowing:

- **`File > Locate in Drive`** opens the containing folder directly. This is the fastest answer to "where is my notebook."
- **Rename it as soon as you save it.** `Copy of 0_Welcome_and_Setup` tells you nothing in November.

**Caution:** open all six Module 1 notebooks, save a copy of each, and you now have six files named `Copy of ...` sitting in one flat folder. By week four that folder is unusable. Make yourself a `Drive / OPIM 5509 /` folder now and drag your copies into it as you go - future you is the one who benefits.

**Remember:** there is a second option on that menu, `File > Save a copy in GitHub`. That one commits to a repository **you** own, so it needs a GitHub account and a one-time permission grant. It cannot write to my repo, so you cannot break anything for the rest of the class by clicking it. For this course, Drive is the simpler habit.

## 4. A ninety-second Python sanity check

If any of the cells below make you pause, that is useful information - not a problem. Flag it in office hours during week one, not week nine.

**On your own:** type these out rather than running them as-is. Muscle memory for syntax is real and it only comes from typing.

````python
# Indexing starts at ZERO. This trips people up all semester, including me.
nums = [10, 20, 30, 40, 50]
print("first element :", nums[0])
print("last element  :", nums[-1])
print("first three   :", nums[0:3])   # up to but NOT including index 3
````

````python
# A dictionary maps keys to values. Keras model configs are basically dictionaries.
config = {"layers": 3, "units": 64, "activation": "relu"}
for key, value in config.items():
    print(f"{key:>12} : {value}")
````

````python
# A function you write once and reuse. We lean on this pattern hard in Module 2 and beyond.
def describe_column(frame, column):
    """Print a compact summary of one numeric column."""
    s = frame[column]
    print(f"{column}: mean={s.mean():.3f}  median={s.median():.3f}  "
          f"min={s.min():.3f}  max={s.max():.3f}")

describe_column(df, "MedInc")
describe_column(df, "MedHouseVal")
````

````python
# numpy arrays and their SHAPE. Shape is the single most important idea in deep learning -
# most of your future errors will be shape errors, so start reading shapes now.
arr = np.array(df[["MedInc", "HouseAge", "AveRooms"]])

print("type :", type(arr))
print("shape:", arr.shape, "-> (rows, columns)")
print("first row:", arr[0])
````
