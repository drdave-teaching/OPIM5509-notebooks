  * Conv2D output size: L - (M-1). 28 with a 5x5 kernel -> 24. Say it as "minus (M minus one)", NOT minus M minus 1.
  * Conv2D params: (M*N*channels + 1 bias) * filters -> (5*5*1+1)*3 = 78.
  * MaxPooling halves it (24->12) and has ZERO trainable params - nothing is learned in pooling, it just takes the max.
  * Second conv: each filter is a 3x3x3 BLOCK. The last 3 is INHERITED - the previous layer handed us 3 feature maps, so every filter must be 3 deep to see all of them. You never type that 3; Keras reads it off the input. The 5 is the only number YOU choose: how many blocks to make. ((3*3*3)+1)*5 = 140.
* Say it twice: depth IN is inherited, count OUT is chosen. L=3 in, F=5 out - different jobs, different numbers, on purpose.
  * Flatten 5*5*5 = 125 -> Dense: 125*256+256 = 32,256 -> output 256*2+2 = 514. Total 32,988.
- HEADS-UP, Keras 3 in Colab: (1) a red "Do not pass an input_shape..." warning prints - harmless, tell students to ignore it;

```
from tensorflow.keras import layers
from tensorflow.keras import models
model = models.Sequential()

# CONV2D_1 - first convolutional 2D layer
# notes: light downsampling, we get to specify the kernel size, and number of feature maps
# the input image stats are given (rows, columns, channels)
# stride is always 1 in our Convolutional Neural Networks
# because stride is 1, there's a ton of overlap... which means our
# outputted feature maps are usually CLOSE but smaller, than their inputs
model.add(layers.Conv2D(3, # number of feature maps we generate
                        (5, 5), # filter/kernel/window size = m,n
                        activation='relu', # activation function
                        input_shape=(28, 28, 1))) #(rows,columns,channels)
                              #channels=1 means BW, channels = 3 means RGB (Red, Green, Blue color)
                              #what happens to param# if channels=3!

# MAXPOOL2D_1 - first max pooling layer
# notes: aggressive downsampling, and we just need to specify kernel size (m,n)
# it's ALWAYS a SQUARE SHAPE (like 2,2 or 5,5... it's given in the code/up to you)
# stride is ALWAYS equal to M (same as kernel size in that layer, like 2,2 or 5,5...)
# because stride is equal to M, there is NO OVERLAP!!! This is why it's aggressive downsampling
model.add(layers.MaxPooling2D((2, 2)))

# CONV2D_2: second convolutional layer
model.add(layers.Conv2D(5, (3, 3), activation='relu')) # S0 - (M-1); 3 maps IN, 5 maps OUT

# MAXPOOL2D_2
model.add(layers.MaxPooling2D((2, 2)))

# FLATTEN: go from arrays from pooling layer to a [1,n] row
# this makes it look like the first row in a dataframe .csv
# the first training row...
model.add(layers.Flatten())

# DENSE_1: this is a hidden layer
model.add(layers.Dense(256, activation='relu'))

# DENSE_2: this is the output node...
# if it's a regression problem, it will have a 'linear' activation function and 1 unit
# if it's a multi-class or binary classification problem, you may have 2 or more nodes...
# and it will use a 'softmax' activation
model.add(layers.Dense(2, activation='softmax'))

model.summary()

# play with the feature maps or filter size!
```

```
# WHERE IS THE BIAS? One bias PER FILTER (per output feature map) - not per pixel!
# The kernel and its bias travel together as the filter slides across the image.
W, b = model.layers[0].get_weights()
print('kernel weights:', W.shape)   # (5, 5, 1, 3) -> an M x N x L brick, one per filter
print('biases:        ', b.shape)   # (3,) -> three filters, three biases. That's it!
```
